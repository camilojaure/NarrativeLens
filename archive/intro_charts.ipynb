# Gráficos — Introducción de Tesis

Genera los tres charts de la intro con estética académica (ggplot-style).  
Los PNG se guardan directamente en la carpeta del proyecto Overleaf para que LaTeX los encuentre.

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
import os

# --- Output path (carpeta del .tex) ---
OUTPUT_DIR = os.path.join(
    os.path.dirname(os.path.abspath('__file__')),
    '../../Tesis de Maestria Management & Analytics/tesis_mma_itba'
)
os.makedirs(OUTPUT_DIR, exist_ok=True)

# --- Paleta académica ---
COLOR_DIGITAL    = '#C0392B'   # rojo oscuro
COLOR_TRAD       = '#E8A838'   # naranja/dorado
COLOR_GRID       = '#CCCCCC'
COLOR_TEXT       = '#2C2C2C'

# --- Estilo base ggplot customizado ---
plt.rcParams.update({
    'axes.facecolor':    '#F5F5F5',
    'figure.facecolor':  'white',
    'axes.edgecolor':    '#AAAAAA',
    'axes.linewidth':    0.8,
    'axes.grid':         True,
    'grid.color':        'white',
    'grid.linewidth':    1.0,
    'axes.spines.top':   False,
    'axes.spines.right': False,
    'font.family':       'sans-serif',
    'font.size':         11,
    'axes.titlesize':    13,
    'axes.labelsize':    11,
    'xtick.labelsize':   10,
    'ytick.labelsize':   10,
    'legend.fontsize':   10,
    'figure.dpi':        150,
})

print('Setup listo. Output dir:', OUTPUT_DIR)
```

## 1. Global Ad Spend 2019–2025

Fuente: Cramer-Flood (2025) y eMarketer (2024). Valores en USD miles de millones (aproximados).

```python
years = [2019, 2020, 2021, 2022, 2023, 2024, 2025]

# USD billions — ajustar si tenés los valores exactos de la fuente
traditional = [320, 265, 295, 285, 275, 270, 255]
digital     = [340, 350, 425, 495, 565, 685, 780]

x = np.arange(len(years))
bar_w = 0.55

fig, ax = plt.subplots(figsize=(8, 5))

b1 = ax.bar(x, traditional, width=bar_w, color=COLOR_TRAD,    label='Traditional Media', zorder=3)
b2 = ax.bar(x, digital,     width=bar_w, color=COLOR_DIGITAL, label='Digital Media',
            bottom=traditional, zorder=3)

ax.set_xticks(x)
ax.set_xticklabels(years)
ax.set_xlabel('Year', color=COLOR_TEXT)
ax.set_ylabel('Spend (USD billions)', color=COLOR_TEXT)
ax.set_title('Global Ad Spend 2019–2025 (USD billions, approximate)', color=COLOR_TEXT, pad=12)
ax.set_ylim(0, 1150)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f'{int(v):,}'))
ax.tick_params(colors=COLOR_TEXT)

ax.legend(loc='upper left', frameon=False)

fig.tight_layout()
fig.savefig(os.path.join(OUTPUT_DIR, 'global_ad_spend.png'), bbox_inches='tight', dpi=200)
plt.show()
print('Guardado: global_ad_spend.png')
```

## 2. Diversidad creativa como "llaves" de micro-segmentos

Diagrama conceptual: múltiples piezas creativas → algoritmo → micro-segmentos de audiencia.

```python
fig, ax = plt.subplots(figsize=(9, 5))
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis('off')
ax.set_facecolor('white')
fig.patch.set_facecolor('white')

# Colores para las creatividades
creative_colors = ['#C0392B', '#E8A838', '#2980B9', '#27AE60', '#8E44AD']
creative_labels = [
    'Formato A\n(Humor)', 'Formato B\n(Evento)', 'Formato C\n(UGC)',
    'Formato D\n(Demo)', 'Formato E\n(Collab)'
]
segment_labels = [
    'Segmento 1\n(18–24, F)', 'Segmento 2\n(25–34, M)',
    'Segmento 3\n(Intención alta)', 'Segmento 4\n(Retargeting)'
]
segment_colors = ['#C0392B', '#E8A838', '#2980B9', '#27AE60']

# Creatividades (izquierda)
y_creatives = np.linspace(2, 8, len(creative_labels))
for i, (yc, label, col) in enumerate(zip(y_creatives, creative_labels, creative_colors)):
    rect = mpatches.FancyBboxPatch((0.3, yc - 0.45), 1.8, 0.9,
                                    boxstyle='round,pad=0.08',
                                    facecolor=col, edgecolor='white',
                                    linewidth=1.5, alpha=0.88)
    ax.add_patch(rect)
    ax.text(1.2, yc, label, ha='center', va='center',
            fontsize=8.5, color='white', fontweight='bold')

# Algoritmo (centro)
algo_box = mpatches.FancyBboxPatch((3.8, 4.0), 2.4, 2.0,
                                    boxstyle='round,pad=0.15',
                                    facecolor='#2C3E50', edgecolor='white',
                                    linewidth=1.5, alpha=0.92)
ax.add_patch(algo_box)
ax.text(5.0, 5.35, 'Algoritmo', ha='center', va='center',
        fontsize=10, color='white', fontweight='bold')
ax.text(5.0, 4.75, 'de distribución', ha='center', va='center',
        fontsize=8.5, color='#BDC3C7')
ax.text(5.0, 4.2, '(TikTok / Meta)', ha='center', va='center',
        fontsize=7.5, color='#95A5A6')

# Micro-segmentos (derecha)
y_segments = np.linspace(2.5, 7.5, len(segment_labels))
for ys, label, col in zip(y_segments, segment_labels, segment_colors):
    circle = plt.Circle((8.5, ys), 0.7, color=col, alpha=0.82)
    ax.add_patch(circle)
    ax.text(8.5, ys, label, ha='center', va='center',
            fontsize=7.5, color='white', fontweight='bold')

# Flechas creatividades → algoritmo
for yc in y_creatives:
    ax.annotate('', xy=(3.85, 5.0), xytext=(2.1, yc),
                arrowprops=dict(arrowstyle='->', color='#7F8C8D',
                                lw=1.2, connectionstyle='arc3,rad=0.0'))

# Flechas algoritmo → segmentos
for ys in y_segments:
    ax.annotate('', xy=(7.8, ys), xytext=(6.2, 5.0),
                arrowprops=dict(arrowstyle='->', color='#7F8C8D',
                                lw=1.2, connectionstyle='arc3,rad=0.0'))

# Etiquetas de sección
ax.text(1.2, 1.2, 'Piezas creativas', ha='center', fontsize=9,
        color='#555555', style='italic')
ax.text(8.5, 1.2, 'Audiencias\nmicro-segmentadas', ha='center', fontsize=9,
        color='#555555', style='italic')

ax.set_title('La diversidad creativa desbloquea micro-segmentos de audiencia',
             fontsize=12, color=COLOR_TEXT, pad=10)

fig.tight_layout()
fig.savefig(os.path.join(OUTPUT_DIR, 'diversidad_llaves.png'), bbox_inches='tight', dpi=200)
plt.show()
print('Guardado: diversidad_llaves.png')
```

## 3. Radar — Creative Diversity Score (CDS)

Ejemplo con CDI = 70.6/100. Ajustar los valores por dimensión si Ready Set tiene un ejemplo canónico.

```python
dimensions = [
    'Formato /\nProducción',
    'Temática /\nConcepto',
    'Funnel /\nSegmentación',
    'Canales /\nUbicaciones',
    'Casting /\nTalent',
    'Rotación /\nRenovación'
]

# Valores de ejemplo que promedian ~70.6 — ajustar si tenés los reales
values = [75, 68, 60, 80, 72, 68]

N = len(dimensions)
angles = np.linspace(0, 2 * np.pi, N, endpoint=False).tolist()

# Cerrar el polígono
values_plot = values + [values[0]]
angles_plot = angles + [angles[0]]

fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(polar=True))
fig.patch.set_facecolor('white')
ax.set_facecolor('#F5F5F5')

# Grilla y rangos
ax.set_ylim(0, 100)
ax.set_yticks([20, 40, 60, 80, 100])
ax.set_yticklabels(['20', '40', '60', '80', '100'], fontsize=7.5, color='#888888')
ax.yaxis.grid(True, color='white', linewidth=1.0)
ax.xaxis.grid(True, color='white', linewidth=1.0)
ax.spines['polar'].set_visible(False)

# Área rellena
ax.fill(angles_plot, values_plot, color=COLOR_DIGITAL, alpha=0.25)
ax.plot(angles_plot, values_plot, color=COLOR_DIGITAL, linewidth=2.0)
ax.scatter(angles, values, color=COLOR_DIGITAL, s=55, zorder=5)

# Etiquetas de dimensión
ax.set_xticks(angles)
ax.set_xticklabels(dimensions, fontsize=9, color=COLOR_TEXT)

# Título con score
ax.set_title(f'Creative Diversity Index: 70.6 / 100',
             fontsize=12, color=COLOR_TEXT, pad=20, fontweight='bold')

fig.tight_layout()
fig.savefig(os.path.join(OUTPUT_DIR, 'radar_cds.png'), bbox_inches='tight', dpi=200)
plt.show()
print('Guardado: radar_cds.png')
```
